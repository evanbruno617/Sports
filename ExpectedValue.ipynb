```python
def getInningRuns(dataframe):
    begInning = 1
    topInning = True
    begHomeRuns = 0
    begAwayRuns = 0
    awayInningRuns = 0
    homeInningRuns = 0
    begIndex = 0
    
    for i in range(len(dataframe)):
    
        if (dataframe.loc[i, 'inning'] != begInning) | (dataframe.loc[i, 'isTopInning'] != topInning):
            
            if (dataframe.loc[(i-1), 'isTopInning'] == True):
                awayInningRuns = dataframe.loc[(i-1), 'awayScore'] - begAwayRuns
                for j in range(begIndex, i, 1):
                    dataframe.loc[j, 'inningRuns'] = awayInningRuns
                
            elif (dataframe.loc[(i-1), 'isTopInning'] == False):
                homeInningRuns = dataframe.loc[(i-1), 'homeScore'] - begHomeRuns
                for j in range(begIndex, i, 1):
                    dataframe.loc[j, 'inningRuns'] = homeInningRuns
            
            begInning = dataframe.loc[i, 'inning'] 
            topInning = dataframe.loc[i, 'isTopInning']
            begHomeRuns = dataframe.loc[i, 'homeScore']
            begAwayRuns = dataframe.loc[i, 'awayScore']
            inningRuns = 0
            begIndex = i
```
